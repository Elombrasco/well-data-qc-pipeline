```python
import pandas as pd 

#loading the data
df = pd.read_excel('../data/Volve production data.xlsx')
print("Shape:", df.shape)
df.head() 
```

```python
# Dimensions du dataset
print("Nombre de lignes:", df.shape[0])
print("Nombre de colonnes:", df.shape[1])

# Infos générales 
print("\nInformations générales sur le dataset:")
print(df.info())

# Valeurs manquantes
print("\nPourcentage de valeurs manquantes par colonne:")    
print((df.isnull().mean() * 100).round(2))  # Affiche le pourcentage de valeurs manquantes par colonne, arrondi à 2 décimales
```

```python
# Quelle période couvre le dataset?
print("\nPériode couverte par le dataset:")
print("Date de début:", df['DATEPRD'].min())
print("Date de fin:", df['DATEPRD'].max())

# Quels puits sont inclus dans le dataset?
print("\nPuits inclus dans le dataset:")    
print(df['NPD_WELL_BORE_NAME'].unique())

# Répartition par types de puits
print("\nRépartition par types de puits:")
print(df['WELL_TYPE'].value_counts())

# Répartition par types de flow
print("\nRépartition par types de flow:")
print(df['FLOW_KIND'].value_counts())
```

```python
# ============================================
# PHASE PROCESS — Nettoyage des données
# ============================================

# 1. Colonnes inutiles pour notre analyse
cols_to_drop = [
    'NPD_WELL_BORE_CODE',  # code numérique redondant avec le nom
    'NPD_FIELD_CODE',      # toujours VOLVE
    'NPD_FACILITY_CODE',   # toujours MÆRSK INSPIRER
    'NPD_FACILITY_NAME',   # toujours MÆRSK INSPIRER
    'NPD_FIELD_NAME',      # toujours VOLVE
    'AVG_CHOKE_UOM'        # unité toujours en %
]

df_clean = df.drop(columns=cols_to_drop)
print("Colonnes après nettoyage :", df_clean.shape[1])
print(df_clean.columns.tolist())
```

```python
#Traitement des valeurs manquantes
print("\n Values manquantes avant traitement:")
print(df_clean.isnull().sum())

# Séparer les puits producteurs et injecteurs
df_prod = df_clean[df_clean['WELL_TYPE'] == 'OP'].copy()  # Puits producteurs 
df_inj = df_clean[df_clean['WELL_TYPE'] == 'WI'].copy()  # Puits injecteurs

# Pour les puits producteurs, remplacer les Nan par 0 
# (pas de production signifie 0, pas de données manquantes)
prod_cols = ['BORE_OIL_VOL', 'BORE_GAS_VOL', 'BORE_WAT_VOL']
df_prod[prod_cols] = df_prod[prod_cols].fillna(0)

# Pour les mesures de pression, remplacer les Nan par la médiane du puits   
pressure_cols = [
    'AVG_DOWNHOLE_PRESSURE', 'AVG_DOWNHOLE_TEMPERATURE', 'AVG_ANNULUS_PRESS',
    'AVG_WHP_P', 'AVG_WHT_P'
]

for col in pressure_cols:
    df_prod[col] = df_prod.groupby('NPD_WELL_BORE_NAME')[col].transform(
        lambda x: x.fillna(x.median())
    )

print("\n Values manquantes après traitement:")
print(df_prod.isnull().sum())
```

```python
df_prod = df_prod.drop(columns=['BORE_WI_VOL'])  # inutile pour les OP

# Les rares NaN restants → on supprime ces lignes
df_prod = df_prod.dropna(subset=[
    'AVG_DOWNHOLE_PRESSURE',
    'AVG_DOWNHOLE_TEMPERATURE',
    'AVG_DP_TUBING',
    'AVG_CHOKE_SIZE_P',
    'DP_CHOKE_SIZE'
])

# Verification finale 
print("Shape après nettoyage:", df_prod.shape)
print("Valeurs nulles:", df_prod.isnull().sum().sum())
print("Puits présents:", df_prod['NPD_WELL_BORE_NAME'].nunique())
print("Periode  couverte:", df_prod['DATEPRD'].min(), "to", df_prod['DATEPRD'].max())
```

```python
# Sauvegarde des données nettoyées
df_prod.to_csv('../data/volve_production_clean.csv', index=False)
print("Données sauvegardées dans data/volve_production_clean.csv")
```

```python
# Vérification des doublons
print("--- DOUBLONS ---")
doublons = df_prod.duplicated().sum()
print(f"Nombre de doublons : {doublons}")

# Vérifier aussi les doublons par puits + date
# Un puits ne peut pas avoir 2 enregistrements le même jour
doublons_cles = df_prod.duplicated(
    subset=['NPD_WELL_BORE_NAME', 'DATEPRD']
).sum()
print(f"Doublons puits + date : {doublons_cles}")
```

```python
# ============================================
# PHASE ANALYZE — Analyse des données
# ============================================

# Ici on pourrait faire des analyses exploratoires plus poussées, comme :
# - Quel puits produit le plus ? Le moins ?
# - Quand la production à t'elle décliné?
# - Quel puits a le plus fort water cut?
# - Y 'a t'il des anomalies de pression? 


import plotly.express as px

# ============================================
# PHASE ANALYZE — Question 1
# Quel puits produit le plus de pétrole ?
# ============================================

oil_by_well = df_prod.groupby('NPD_WELL_BORE_NAME')['BORE_OIL_VOL']\
                     .sum()\
                     .reset_index()\
                     .sort_values('BORE_OIL_VOL', ascending=False)

oil_by_well.columns = ['Puits', 'Volume total huile (Sm3)']

fig = px.bar(
    oil_by_well,
    x='Puits',
    y='Volume total huile (Sm3)',
    title='Production totale de pétrole par puits — Champ Volve',
    color='Volume total huile (Sm3)',
    color_continuous_scale='Oranges',
    text_auto='.2s'
)

fig.update_layout(
    plot_bgcolor='white',
    title_font_size=16
)

fig.show()
```

 2 puits sur 5 représentent 87% de la production totale du champ. 

```python
# Question 2 — Déclin de production dans le temps
oil_by_date = df_prod.groupby('DATEPRD')['BORE_OIL_VOL']\
                     .sum()\
                     .reset_index()

fig = px.line(
    oil_by_date,
    x='DATEPRD',
    y='BORE_OIL_VOL',
    title='Évolution de la production journalière — Champ Volve',
    labels={
        'DATEPRD': 'Date',
        'BORE_OIL_VOL': 'Production huile (Sm3/jour)'
    },
    color_discrete_sequence=['#E07B00']
)

fig.update_layout(plot_bgcolor='white')
fig.show()
```

 ##### - 2008-2010  → Phase montée en production (peak à ~9k Sm3/jour)
##### - 2010-2012  → Déclin naturel rapide du champ
##### - Chutes à 0 → Arrêts puits (maintenance, interventions)
##### - 2012-2016  → Déclin lent et régulier jusqu'à l'abandon du champ

Le champ Volve a atteint son pic en 2009-2010 puis a décliné naturellement — c'est le comportement classique d'un réservoir en déplétion.

```python
#  Water Cut par puits
# Water Cut = % eau dans la production totale liquide

df_prod['WATER_CUT'] = (
    df_prod['BORE_WAT_VOL'] /
    (df_prod['BORE_OIL_VOL'] + df_prod['BORE_WAT_VOL'])
) * 100

water_cut = df_prod.groupby('NPD_WELL_BORE_NAME')['WATER_CUT']\
                   .mean()\
                   .reset_index()\
                   .sort_values('WATER_CUT', ascending=False)

fig = px.bar(
    water_cut,
    x='NPD_WELL_BORE_NAME',
    y='WATER_CUT',
    title='Water Cut moyen par puits — Champ Volve',
    labels={
        'NPD_WELL_BORE_NAME': 'Puits',
        'WATER_CUT': 'Water Cut moyen (%)'
    },
    color='WATER_CUT',
    color_continuous_scale='Blues',
    text_auto='.1f'
)

fig.update_layout(plot_bgcolor='white')
fig.show()
```

##### - 15/9-F-14  → 65.2% eau  — puits très watérisé, coût de traitement élevé
##### - 15/9-F-12  → 59.6% eau  — malgré d'être le 1er producteur !
##### - 15/9-F-1 C → 51.1% eau  — plus d'eau que de pétrole
##### - 15/9-F-11  → 38.4% eau  — acceptable
##### - 15/9-F-15D → 19.3% eau  — le plus propre du champ

Le puits 15/9-F-14 produit beaucoup mais avec 65% d'eau — ce qui signifie des coûts de séparation et traitement très élevés

```python
# Anomalies de pression par puits
fig = px.box(
    df_prod,
    x='NPD_WELL_BORE_NAME',
    y='AVG_DOWNHOLE_PRESSURE',
    title='Distribution de la pression fond de puits par puits',
    labels={
        'NPD_WELL_BORE_NAME': 'Puits',
        'AVG_DOWNHOLE_PRESSURE': 'Pression fond (bar)'
    },
    color='NPD_WELL_BORE_NAME',
    points='outliers'
)

fig.update_layout(
    plot_bgcolor='white',
    showlegend=False
)

fig.show()
```

##### - 15/9-F-12  → Boîte énorme (0 à 300 bar) — pression très instable
##### - 15/9-F-14  → Beaucoup d'outliers bas — chutes de pression anormales  
##### - 15/9-F-15D → Outliers à 390 bar — pics de surpression suspects
##### - 15/9-F-1 C → Distribution serrée — puits stable
##### - 15/9-F-11  → Distribution serrée — puits stable

Le puits 15/9-F-12 malgré être le top producteur montre une pression très erratique — signe probable d'interventions multiples ou de problèmes d'intégrité du puits.
